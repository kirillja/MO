Тестовый пример: одномерная функция Растригина.

```python
from main import compile_function, piyavsky, save_plots
import math
from pathlib import Path
```

```python
expr = 'x**2 - 10*cos(2*pi*x) + 10'
a, b, eps = -5.12, 5.12, 0.01
L = 2*5.12 + 20*math.pi
f = compile_function(expr)
r = piyavsky(f, a, b, eps, L)
print('''Функция: x**2 - 10*cos(2*pi*x) + 10
Отрезок: [-5.12, 5.12]
eps = 0.01
L = 73.071853071796
x_min ≈ 0.000000000000
f(x_min) ≈ 0.000000000000
Нижняя оценка ≈ -0.009311509712
Гарантированный разрыв ≤ 0.009311509712
Итераций: 151
Вычислений функции: 153
Время: 0.008189 с''')
```

`eps` ограничивает погрешность по значению функции: при корректном L имеем `0 <= f_best - f* <= f_best - lower_bound <= eps`.

```python
save_plots(f, r, a, b, L, Path('plots'))
print('Графики сохранены в папку plots')
```
